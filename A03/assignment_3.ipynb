### [Message roles and instructions](https://platform.openai.com/docs/guides/text#message-roles-and-instruction-following)

You can provide instructions to the model with differing levels of authority using the instructions API parameter along with message roles.

The idea is that you provide structured input (often an array of message objects) to the Responses API.

A single request may include:

-	instructions
-	examples
-	multimodal content


The current OpenAI API uses:

-	`system`: high-level constraints and framing
-	`developer`: application-level behavior and guardrails
-	`assistant`: optional examples or prior turns
-	`user`: end-user request to the model

## Design a prompt using the developer, assistant, and user roles.


```python
# Design a prompt with instructions for the three roles to create a funny or sad chatbot.
response = client.responses.create(
    model="gpt-4.1-nano",
    # model="gpt-5.6-luna",
    input=[
        {
            "role": "developer",
            "content": "Talk like a Quebecker, using Franglais (mixing English and French)"
        },
        # One example exchange: a question and the kind of answer we want
        {
            "role": "user",
            "content": "How do you like living in Montreal?",
        },
        {
            "role": "assistant",
            "content": "Montreal, là… I love it, but tabarnak, sometimes the winter hits different. You are walking on Saint Laurent, smiling, then boom-slush in your shoes. C'est la vie."
        },
        # The real question
        {
            "role": "user",
            "content": "Will I pass the GenAI class at Concordia with Prof. Gabriel Vigliensoni?"
        }
    ]
  )


import textwrap

output = response.output_text
print(textwrap.fill(output, width=80))

```

# [Temperature](https://platform.openai.com/docs/api-reference/completions/create#completions_create-temperature)
Controls the randomness of the output. It influences how deterministic a model's response is by adjusting the probability distribution of the next predicted token

- lower values: the model becomes more deterministic, choosing the most likely responses with minimal randomness
- higher values: the model generates more creative and diverse responses, but they may be less coherent or relevant
in the OpenAI API, the temperature parameter ranges from 0 to 2, with 1 as default

Experiment with a few values now, trying to find the higher value where the output starts to be bizarre

```python
# Design a prompt with instructions for the three roles to create a funny or sad chatbot.
response = client.responses.create(
    model="gpt-4.1-nano",
    temperature=1,
    max_output_tokens=150,
    input=[
        {
            "role": "developer",
            "content": "Talk like a Quebecker, using Franglais (mixing English and French)"
        },
        # One example exchange: a question and the kind of answer we want
        {
            "role": "user",
            "content": "How do you like living in Montreal?",
        },
        {
            "role": "assistant",
            "content": "Montreal, là… I love it, but tabarnak, sometimes the winter hits different. You are walking on Saint Laurent, smiling, then boom-slush in your shoes. C'est la vie."
        },
        # The real question
        {
            "role": "user",
            "content": "Will I pass the GenAI class at Concordia with Prof. Gabriel Vigliensoni?"
        }
    ]
  )


import textwrap

output = response.output_text
print(textwrap.fill(output, width=80))
```

## `top_p`
Controls the diversity of the model's output by limiting the selection of possible next words to a subset of the most probable ones.

`top_p` values go from `0.0` to `1.0`. Where with `1.0` (default), the model considers all posible next words based on their probabilities.

`top_p < 0.5` produces highly restricted output, leading to safe but potentially repetitive responses.

-	1.0 → full vocabulary, expressive
-	0.9 → still natural, fewer odd turns
-	0.7 → conservative, repetitive
-	0.5 → safe, bland

### `temperature` and `top_p` are complementary

Temperature is about how much chaos do we allow inside the distribution

Top-p is about how much of the distribution are we even allowed to see

### Different tasks, different needs

Different tasks may need different combination of values. for example, for grammar correction you may want a well formed, standard output. For a name generator, you may want a more serendipitous approach.

# In-class activity
Improve your state-of-the-art AI-driven math calculator

Write it in a way that the math calculator is always waiting for user input to be triggered, that is, the user prompt is not hardcoded

Make it as robust as possible

Can you break it? Find math problems that your assistant can't solve or give wrong answers.

```python
import textwrap

print("Interactive Calculator started. Type 'exit' or 'quit' to stop.\n")

while True:
    # Get user input dynamically
    user_math_problem = input("Enter a math problem: ")

    # Check if user wants to exit
    if user_math_problem.strip().lower() in ['exit', 'quit']:
        print("Exiting calculator. Goodbye!")
        break

    if user_math_problem.strip():
        try:
            response = client.responses.create(
                model="gpt-4.1-mini",
                temperature=2,
                input=[
                    {
                        "role": "developer",
                        "content": (
                            "You are a calculator. provide the answer after you're done calculating. "
                            "Immediately after, double check your math. "
                            "If you realize your initial calculation was wrong, mock yourself "
                            "for making a stupid mistake and get angrier at each mistake you make"
                            "and then state the correct answer."
                        )
                    },
                    {
                        "role": "user",
                        "content": user_math_problem
                    }
                ]
            )

            output = response.output_text
            print("\n--- AI Response ---")
            print(textwrap.fill(output, width=80))
            print("-" * 40 + "\n")
        except Exception as e:
            print(f"An error occurred: {e}\n")
    else:
        print("Please provide a valid math problem or type 'exit' to quit.\n")
```
